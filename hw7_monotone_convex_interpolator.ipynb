### Include an IFR curve plot and briefly explain how your method ensures monotonicity and convexity.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from fixedincomelib import qfCreate1DInterpolator
from fixedincomelib.date import Date, accrued

value_date = Date("2026-01-02")
node_times = [accrued(value_date, Date(d)) for d in node_dates]
sofr_values = sofr_quotes["values"].to_numpy()

pcp_curve = qfCreate1DInterpolator(
    node_times, sofr_values, "PIECEWISE_CONSTANT_LEFT_CONTINUOUS", "FLAT",
)
mc_curve = qfCreate1DInterpolator(
    node_times, sofr_values, "MONOTONE_CONVEX", "FLAT",
)

grid = np.linspace(0.0, node_times[-1] * 1.15, 1200)
pcp_vals = [pcp_curve.interpolate(x) for x in grid]
mc_vals = [mc_curve.interpolate(x) for x in grid]

plt.figure(figsize=(9, 5))
plt.plot(grid, pcp_vals, label="Piecewise constant (left-continuous)", linewidth=1.5)
plt.plot(grid, mc_vals, label="Monotone convex (mine)", linewidth=1.8)
plt.scatter(node_times, sofr_values, color="black", zorder=5, label="Supplied SOFR-1B IFR nodes")
plt.xlabel("Time from value date (years)")
plt.ylabel("Instantaneous forward rate")
plt.title("SOFR-1B instantaneous forward curve")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

On each node interval \([t_k, t_{k+1}]\) the SOFR-1B node values are point samples of the instantaneous
forward curve, so the interpolant must reproduce them exactly; there is no separate bucket-average quote
to match the way Hagan-West's original three-constraint quadratic does. I therefore fit the quadratic
\(q(x) = v_k + b x + c x^2\), \(x=(t-t_k)/(t_{k+1}-t_k)\in[0,1]\), through the two exact node values and
fix the remaining degree of freedom with a Hagan-West/PCHIP-style shape-preserving derivative estimate
\(d_k\) at the left node (weighted harmonic mean of the two neighbouring secant slopes \(S_{k-1}, S_k\), set
to zero at a local extremum, one-sided at the ends).

\(d_k\) is then clamped against the segment's own secant \(S_k=(v_{k+1}-v_k)/(t_{k+1}-t_k)\) so that
\(0\le d_k\le 2S_k\) (or \(2S_k\le d_k\le 0\) if \(S_k<0\)). Since \(q'(x)\) is linear in \(x\), same-signed
endpoint slopes \(q'(0)=d_k\) and \(q'(1)=2S_k-d_k\) mean \(q'\) never changes sign on \([0,1]\) — the segment
is **monotone**. Because \(q\) is quadratic, \(q''=2c\) is constant on the segment, so curvature never
changes sign either — the segment is **convex (or concave) throughout, with no interior inflection**, i.e.
"monotone convex" on every node interval. The curve is continuous (both neighbouring segments pin down the
same node value) though, as in Hagan-West's own construction, the first derivative can kink at a node.
Outside the node range the curve extrapolates flat at the boundary value, and a single-node curve (the
funding IFR) is treated as a constant.

The plot above shows this clearly against the piecewise-constant convention: the monotone-convex curve
threads through the same six node values with no overshoot, stays monotone wherever the input itself is
monotone, and bends smoothly instead of jumping at each knot.
